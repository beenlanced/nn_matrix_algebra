## Self-Attention

Attention helps keep track of the relationships between words which pairs well with the positional input information. Self-attention works by seeing how similar each word is to all of other words in the sentence, including itself. In other words, calculating the similarity between words. 

<p>
  <img alt="Self-Attention" src="images/attention_1.png" width="300" height="200"/>
</p>

[img src](https://www.youtube.com/watch?v=KphmOJnLAdI&list=PLblh5JKOoLUIxGDQs4LFFD--41Vzf-ME1&index=25)

Three key matrices **Q, K, and V** are used here to determine the relevance of each word to the others in the context of the input sentence.

* **Q-Queries**: represents the processed information of the current word

* **K-Keys**: contains the processed information of all words in the sentence

* **V-Values**: holds the raw information of all words

Calculating each one looks like the diagram below where random weights are initially assigned and as training of the model occurs the weights would be updated according to loss measurements. The weights include: 0.17, -0.51, 1.36, etc.

##### Calculating the SoftMax Portion
Now Let's look at completing the **SoftMax** calculation portion of the Attention Equation. **SoftMax** converts a tuple of K real numbers into a probability distribution of K possible outcomes. It is a generalization of the logistic function to multiple dimensions, and is used in multinomial logistic regression. The softmax function is often used as the last activation function of a neural network to normalize the output of a network to a probability distribution over predicted output classes.[src](https://en.wikipedia.org/wiki/Softmax_function)

The general idea is to train the model so that the input tokens in English

``` <SOS> Let's go ```

Get translated into the output tokens for the Spanish translation

``` vamos <EOS> ```

Here, `<SOS>` refers to the Start of Sequence and `<EOS>` is the End of Sequence. They are special tokens that are used to indicate the beginning and end of input sequences. The SOS token signals the start of the the input for the decoder, while the EOS token indicates when the sequence is finished -- tells the Decoder to stop generating output.
